```python
rangeScale = 3

btrain,l1 = datasets.make_moons(100000,noise=0.02)
btest,l2 = datasets.make_moons(20000,noise=0.02)

dtrain,l2 = datasets.make_moons(100000,noise=0.02)
dtest,l2 = datasets.make_moons(20000,noise=0.02)
dtrain[:,1] *= 0.75
dtest[:,1] *= 0.75

plt.scatter(btrain[:,0],btrain[:,1],s=2)
plt.scatter(dtrain[:,0],dtrain[:,1],s=2)
```

```python
import torch
import torch.optim as optim
import torch.nn as nn
import torch.nn.init as init
from tqdm import tqdm
import torch.utils.data as utils

btrain,l1 = datasets.make_moons(100000,noise=0.02)
btest,l2 = datasets.make_moons(20000,noise=0.02)
mins = btrain.min(axis=0)
maxes = btrain.max(axis=0)
means = btrain.mean(axis=0)
stds = btrain.std(axis=0)
#btrain = 6*((btrain-mins)/(maxes-mins)-0.5)
#btest = 6*((btest-mins)/(maxes-mins)-0.5)
btrain=(btrain-means)/stds
btest=(btest-means)/stds

device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
torch.set_default_tensor_type('torch.cuda.FloatTensor') if torch.cuda.is_available() else print ('cpu')

kwargs = {"flow_type":"NSRATQUAD","tail_bound":3.5,"hidden_features":120,
          "num_layers":1,"num_bins":60,"num_blocks_per_layer":10,"tails":"linear"}
flow_b = qrnn.make_flow(2,0,kwargs,perm=True).to(device)
bs = 128

btrain_dset = utils.TensorDataset(torch.tensor(btrain,dtype=torch.float32,device=device))
btrain_loader = qrnn.InfiniteLoader(dataset=btrain_dset,batch_size=bs,shuffle=True,
                                    generator=torch.Generator(device='cuda'))

optimizer = optim.Adam(flow_b.parameters(),lr=5e-4)
n_iter = 4000
losses = []
tbar = tqdm(range(n_iter))
min_loss = 1e8
patience_count=0
for i in tbar:
    x, = next(btrain_loader)
    optimizer.zero_grad()
    loss = -flow_b.log_prob(inputs=x)[0].mean()
    loss.backward()
    optimizer.step()
    if loss.item() < min_loss:
        min_loss = loss.item()
        torch.save(flow_b.state_dict(),"random_testing/flow_b.pt")
    losses.append(loss.item())
    if loss.item() == min_loss:
        patience_count=0
        s = 'Loss: {0}, p = {1}'.format(loss.item(),patience_count)
        tbar.set_description(s)
    else:
        patience_count += 1
        s = 'Loss: {0}, p = {1}'.format(min_loss,patience_count)
        tbar.set_description(s)
flow_b.load_state_dict(torch.load("random_testing/flow_b.pt"))
flow_b.eval()
torch.cuda.empty_cache()
plt.figure(figsize=(8,6))
w = int(btrain.shape[0]/(5*bs))
smooth = np.convolve(np.ones(w),losses,mode='valid')/w
xvals = np.linspace(0,len(losses),len(smooth))
plt.plot(xvals,smooth)
```

```python
samples = flow_b.sample(20000).detach().cpu().numpy()
plt.figure(figsize=(8,6))
plt.scatter(btest[:,0],btest[:,1],s=2)
plt.scatter(samples[:,0],samples[:,1],s=2)

qrnn.plotPair(btest[:,0],samples[:,0])
qrnn.plotPair(btest[:,1],samples[:,1])
```
