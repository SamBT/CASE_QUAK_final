```python
import numpy as np
import scipy.stats as stats
```

```python
for i in range(1,15):
    samples = stats.multivariate_normal.rvs(mean=np.zeros(i),cov=np.eye(i),size=10000).reshape(-1,i)
    log_prob = -np.mean(np.log(stats.multivariate_normal.pdf(samples,mean=np.zeros(i),cov=np.eye(i))))
    print(log_prob)
```
