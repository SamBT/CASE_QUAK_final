# Making Mjj-flattened BKG sample

```python
mjj = hf.loadData("BKG",["mjj"],mode="all",dEta_cut=-1)[:,0]

step = 10
bins = np.arange(1200,8000+step,step)
counts,bins = np.histogram(mjj,bins=bins)
upperLims = bins[1:]
possibleWithUniform = []
for i,ct in enumerate(counts):
    nbins = i+1
    ntot = nbins*ct + np.sum(counts[i+1:])
    possibleWithUniform.append(ntot)
possibleWithUniform = np.array(possibleWithUniform)
plt.plot(0.5*(bins[1:]+bins[:-1]),possibleWithUniform)
plt.yscale('log')
plt.grid()
mMax = upperLims[np.max(np.where((possibleWithUniform-112000)>0)[0])]
print(f"Can Flatten out to {mMax}")

# sample uniformly from bulk
bins = np.arange(1200,mMax+step,step)
nBins = len(bins)-1
counts,bulkBins = np.histogram(mjj,bins)
bulkBins = np.digitize(mjj,bins)
nToSample = np.min(counts)
sampleInds = np.concatenate([(bulkBins==i).nonzero()[0][:nToSample] for i in range(1,nBins+1)],axis=0)
tailInds = (mjj>=mMax).nonzero()[0]
flatInds = np.concatenate((sampleInds,tailInds),axis=0)
np.random.shuffle(flatInds)
print(mMax)
del mjj

variables = ['jet1_rho','jet1_mass','jet1_tau21','jet1_tau32','jet1_tau43','jet1_tauS','jet1_logTauS','jet1_pb','jet1_pb2','jet1_npf','jet1_lsf3','jet1_btagscore',
             'jet2_rho','jet2_mass','jet2_tau21','jet2_tau32','jet2_tau43','jet2_tauS','jet2_logTauS','jet2_pb','jet2_pb2','jet2_npf','jet2_lsf3','jet2_btagscore']

#with h5py.File("input_h5s/BKG/BKG_mjjFlat_nodEtaCut.h5","w") as f:
#    for v in variables:
#        data = hf.loadData("BKG",[v],mode="all",dEta_cut=-1)[:,0]
#        sample = data[flatInds]
#        f.create_dataset(v,data=sample)
#        del data, sample
```

# Computing min/maxes for all training variables

```python
from samples import data_merged_dir, bkg_merged_dir, signal_merged_dir
import helper_functions as hf
from tqdm import tqdm
import json
signals = [signal_merged_dir+s for s in os.listdir(signal_merged_dir) if ".h5" in s]
bkgs = [bkg_merged_dir+s for s in os.listdir(bkg_merged_dir) if ".h5" in s]
datas = [data_merged_dir+s for s in os.listdir(data_merged_dir) if ".h5" in s]
allFiles = signals+bkgs+datas

variables = ['jet1_rho','jet1_pt','jet1_eta','jet1_phi','jet1_mass','jet1_tau21','jet1_tau32','jet1_tau43','jet1_tauS','jet1_logTauS','jet1_pb','jet1_pb2','jet1_npf','jet1_lsf3','jet1_btagscore',
             'jet2_rho','jet2_pt','jet2_eta','jet2_phi','jet2_mass','jet2_tau21','jet2_tau32','jet2_tau43','jet2_tauS','jet2_logTauS','jet2_pb','jet2_pb2','jet2_npf','jet2_lsf3','jet2_btagscore',
             'mjj','dEta','signed_dEta']

mins = {v:99999 for v in variables}
maxes = {v:-99999 for v in variables}
for file in tqdm(allFiles):
    with h5py.File(file,"r") as f:
        for v in variables:
            d = hf.getVariable(f,v)
            dmax = d.max()
            dmin = d.min()
            if dmax > maxes[v]:
                maxes[v] = dmax
            if dmin < mins[v]:
                mins[v] = dmin
                
mins = {v:float(mins[v]) for v in mins.keys()}
maxes = {v:float(maxes[v]) for v in maxes.keys()}

output = {
    "mins":mins,
    "maxes":maxes
}
with open("input_h5s/trainVarMinMaxes.json","w") as f:
    json.dump(output,f,indent=4)
```

# Make plot comparing Mjj-flattened background to all background

```python
variables = ['jet1_rho','jet1_tau21','jet1_tau32','jet1_tau43','jet1_logTauS','jet1_pb','jet1_npf',
             'jet2_rho','jet2_tau21','jet2_tau32','jet2_tau43','jet2_logTauS','jet2_pb','jet2_npf']
from helper_functions import varTitle_map
from trainingTools import loadMeans
bkg = hf.loadData("BKG",variables,mode="all",dEta_cut=-1)
bkg_flat = hf.loadData("BKG_mjjFlat",variables,mode="all",dEta_cut=-1)
mean,std = loadMeans("BKG",variables)
bkg = (bkg-mean)/std
bkg_flat = (bkg_flat-mean)/std
```

```python
plt.figure(figsize=(28,8))
for i in range(bkg.shape[1]):
    plt.subplot(2,7,i+1)
    h1,bins,_ = plt.hist(bkg[:,i],bins=50,histtype='step',density=True,label="All Events")
    h2 = plt.hist(bkg_flat[:,i],bins=bins,histtype='step',density=True,label=r"$M_{jj}$-flattened")
    if i == 0 or i == 7:
        plt.legend(loc='upper right')
    plt.xlabel(varTitle_map[variables[i]])
    plt.yscale('log')
plt.suptitle("Background Monte Carlo")
plt.tight_layout()
plt.savefig("background_mjjFlatVsNormal.pdf")
```
