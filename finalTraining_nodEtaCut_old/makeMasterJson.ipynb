```python
%load_ext autoreload
%autoreload 2
import sys
sys.path.append("/home/sambt/CASE_final/CASE_QUAK_final/commonTools/")
import helper_functions as hf
from samples import bkg_merged_dir, signal_merged_dir, data_merged_dir, signal_map
import numpy as np
import h5py
from tqdm import tqdm
import json
import os
```

```python
masterJson = {}
effs = [
    "d_eta_eff",
    "preselection_eff",
    "preselection_eff_JER_down",
    "preselection_eff_JER_up",
    "preselection_eff_JES_down",
    "preselection_eff_JES_up"
]
for sig_pt in tqdm(signal_map.keys()):
    if 'RSG' in sig_pt:
        continue
    masterJson[sig_pt] = {}
    with h5py.File(signal_merged_dir+signal_map[sig_pt],"r") as f:
        for eff in effs:
            masterJson[sig_pt][eff] = f[eff][()][0]
n_bkg_tot = 0
n_bkg_pass_dEta = 0
bkg_files = [bkg_merged_dir+f for f in os.listdir(bkg_merged_dir) if ".h5" in f]
for bf in tqdm(bkg_files):
    with h5py.File(bf,"r") as f:
        dEta = f["jet_kinematics"][:,1]
        n_bkg_tot += len(dEta)
        n_bkg_pass_dEta += np.count_nonzero(np.abs(dEta) < 1.3)
masterJson["BKG"] = {}
masterJson["BKG"]["n_tot"] = n_bkg_tot
masterJson["BKG"]["n_sel"] = n_bkg_pass_dEta
masterJson["BKG"]["d_eta_eff"] = float(n_bkg_pass_dEta/n_bkg_tot)

with open("masterJson.json","w") as f:
    json.dump(masterJson,f,indent=4)
```
