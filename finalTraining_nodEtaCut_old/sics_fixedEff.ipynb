```python
%load_ext autoreload
%autoreload 2
import numpy as np
import matplotlib.pyplot as plt
from quakSpaceTools import sic_fixedEff
```

```python
reducs = ['SumProb','signedL5','l2','chi2z','chi2sf']
sig_sets = [
    ["M80-80","M80-170","M80-400","M170-170","M170-400","M400-400"],
    ["M25-25","M25-80","M25-170","M25-400","M80-80","M80-170","M80-400","M170-170","M170-400","M400-400"],
    ["XYY3000_Y170_Yp170","Wp3000_B400_T170","RSG3000","Qstar2000_W400","Wkk3000_R400"],
]
sig_aliases = ["MassBin_no25","MassBin_all","fiveSigs"]
bases = ["lossEvals_pcaDecorr_boxCox","lossEvals_linearDecorr_boxCox","lossEvals"]
bkg_axes = ["BKG_mjjFlat","DATA_mjjFlat"]
eff = 0.2
```

```python
for (sig_axes,sig_alias) in zip(sig_sets,sig_aliases):
    for bkg_axis in bkg_axes:
        for base in bases:
            sic_fixedEff(bkg_axis,sig_axes,sig_alias,reducs,"XYY3000_Y170_Yp170",eff=eff,decorr=True,base=base)
            sic_fixedEff(bkg_axis,sig_axes,sig_alias,reducs,"XYY3000_Y170_Yp170",eff=eff,decorr=False,base=base)
            sic_fixedEff(bkg_axis,sig_axes,sig_alias,reducs,"Wp3000_B400_T170",eff=eff,decorr=True,base=base)
            sic_fixedEff(bkg_axis,sig_axes,sig_alias,reducs,"Wp3000_B400_T170",eff=eff,decorr=False,base=base)
```
